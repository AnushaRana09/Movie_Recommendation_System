```python
movies["features"] = movies["genre"] + " " + movies["description"]

print("Features created successfully!")
print(movies[["title", "features"]])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(movies["features"])

print("TF-IDF vectorization completed!")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)

print("Similarity calculation completed!")
print("Similarity matrix shape:", similarity.shape)
```

```python
def recommend_movies(movie_title, num_recommendations=5):

    movie_title = movie_title.strip().lower()
    titles = movies["title"].str.lower()

    if movie_title not in titles.values:
        return []

    movie_index = titles[titles == movie_title].index[0]

    similarity_scores = list(enumerate(similarity[movie_index]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for index, score in similarity_scores:

        if index == movie_index:
            continue

        recommendations.append(
            (movies.loc[index, "title"], score)
        )

        if len(recommendations) == num_recommendations:
            break

    return recommendations
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

title = widgets.HTML(
    value="""
    <div style="text-align:center; padding:20px;">
        <h1 style="color:#1f2937; margin-bottom:5px;">
            Movie Recommendation System
        </h1>
        <p style="color:#6b7280; font-size:16px;">
            Find movies similar to your favorite films
        </p>
    </div>
    """
)

search_box = widgets.Text(
    placeholder="Type a movie name...",
    description="Movie:",
    style={"description_width": "70px"},
    layout=widgets.Layout(width="500px")
)

recommend_button = widgets.Button(
    description="Recommend Movies",
    button_style="primary",
    layout=widgets.Layout(width="200px", height="40px")
)

output = widgets.Output()


def show_recommendations(button):

    with output:
        clear_output()

        entered_movie = search_box.value.strip().lower()

        if not entered_movie:
            print("Please enter a movie name.")
            return

        matching_movies = movies[
            movies["title"].str.lower().str.contains(
                entered_movie,
                na=False
            )
        ]

        if matching_movies.empty:
            print("Movie not found in the dataset.")
            return

        if len(matching_movies) > 1:

            print("Multiple movies found. Please type a more specific name:")

            for movie in matching_movies["title"]:
                print("-", movie)

            return

        selected_movie = matching_movies.iloc[0]["title"]

        movie_info = matching_movies.iloc[0]

        details = widgets.HTML(
            value=f"""
            <div style="
                border:1px solid #d1d5db;
                border-radius:10px;
                padding:18px;
                margin-top:20px;
                width:460px;
                background:#f3f4f6;
            ">
                <h2 style="margin:0 0 10px 0; color:#111827;">
                    {movie_info["title"]}
                </h2>

                <p style="margin:5px 0;">
                    <b>Genre:</b> {movie_info["genre"]}
                </p>

                <p style="margin:8px 0; color:#4b5563;">
                    {movie_info["description"]}
                </p>
            </div>
            """
        )

        display(details)

        recommendations = recommend_movies(
            selected_movie,
            num_recommendations=5
        )

        heading = widgets.HTML(
            value="""
            <h2 style="color:#1f2937; margin-top:25px;">
                Recommended Movies
            </h2>
            """
        )

        display(heading)

        for i, (movie_title, score) in enumerate(recommendations, 1):

            card = widgets.HTML(
                value=f"""
                <div style="
                    border:1px solid #d1d5db;
                    border-radius:10px;
                    padding:15px;
                    margin:8px 0;
                    width:460px;
                    background:#ffffff;
                ">
                    <h3 style="margin:0 0 6px 0; color:#111827;">
                        {i}. {movie_title}
                    </h3>

                    <p style="margin:0; color:#6b7280;">
                        Similarity Score: {score:.2f}
                    </p>
                </div>
                """
            )

            display(card)


recommend_button.on_click(show_recommendations)

display(title)
display(search_box)
display(recommend_button)
display(output)
```
